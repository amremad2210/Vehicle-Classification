```python
from pathlib import Path
import shutil

# Adjust if your notebook is in /notebooks
PROJECT_ROOT = Path.cwd().parent
RAW_DIR = PROJECT_ROOT / "data" / "raw"
CLEAN_DIR = PROJECT_ROOT / "data" / "cleaned"

EXCLUDE = {
    "reject",
    "class_12_six_axle_multi_trailer",  # exclude due to too few samples
}

IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

def count_images(folder: Path) -> int:
    return sum(1 for p in folder.glob("*") if p.is_file() and p.suffix.lower() in IMG_EXTS)

# 1) Create cleaned dir
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

# 2) Copy class folders
copied_summary = []
for class_folder in sorted([p for p in RAW_DIR.iterdir() if p.is_dir()]):
    name = class_folder.name
    if name in EXCLUDE:
        continue

    dst = CLEAN_DIR / name
    dst.mkdir(parents=True, exist_ok=True)

    # copy files (flat folder)
    n = 0
    for img in class_folder.iterdir():
        if img.is_file() and img.suffix.lower() in IMG_EXTS:
            shutil.copy2(img, dst / img.name)
            n += 1

    copied_summary.append((name, n))

print("Copied to:", CLEAN_DIR)
print("\nPer-class counts in CLEANED:")
total = 0
for name, n in copied_summary:
    total += n
    print(f"{name}: {n}")
print("TOTAL:", total)
```
