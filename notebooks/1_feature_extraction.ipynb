```python
# Cell 1: Imports and Setup
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

# Add src to path
PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT / "src"))

# IMPORTANT: reload feature_extractor in case it was edited after being imported
import importlib
import feature_extractor
importlib.reload(feature_extractor)
from feature_extractor import VehicleFeatureExtractor, extract_features_from_directory

print(f"Project root: {PROJECT_ROOT}")
print(f"Python version: {sys.version}")
```

```python
# Cell 2: Define Paths
DATA_DIR = PROJECT_ROOT / "data" / "splits"
TRAIN_DIR = DATA_DIR / "train"
VAL_DIR = DATA_DIR / "val"
TEST_DIR = DATA_DIR / "test"

# Output directory for extracted features
FEATURES_DIR = PROJECT_ROOT / "data" / "features"
FEATURES_DIR.mkdir(exist_ok=True, parents=True)

print(f"Train directory: {TRAIN_DIR}")
print(f"Val directory: {VAL_DIR}")
print(f"Test directory: {TEST_DIR}")
print(f"\nFeatures will be saved to: {FEATURES_DIR}")
```

```python
# Cell 3: Initialize Feature Extractor
extractor = VehicleFeatureExtractor(
    img_size=(224, 224),
    hog_orientations=9,
    hog_pixels_per_cell=(8, 8),
    hog_cells_per_block=(2, 2),
    color_bins=32,
    lbp_points=24,
    lbp_radius=3
)

extractor.print_feature_summary()
```

```python
# Cell 4: Test Feature Extraction on Single Image
# Get a sample image
sample_images = list(TRAIN_DIR.glob('class_01_motorcycle/*.jpg'))
if not sample_images:
    sample_images = list(TRAIN_DIR.glob('class_01_motorcycle/*.png'))

if sample_images:
    sample_img = sample_images[0]
    print(f"Testing on: {sample_img.name}")
    
    # Extract features
    features = extractor.extract_all_features(sample_img)
    
    print(f"\nExtracted {len(features)} features")
    print(f"Feature vector shape: {features.shape}")
    print(f"\nFirst 10 features: {features[:10]}")
    print(f"Feature statistics:")
    print(f"  Min: {features.min():.4f}")
    print(f"  Max: {features.max():.4f}")
    print(f"  Mean: {features.mean():.4f}")
    print(f"  Std: {features.std():.4f}")
else:
    print("No sample images found!")
```

```python
# Cell 5: Extract Features from Training Set
print("="*70)
print("EXTRACTING TRAINING SET FEATURES")
print("="*70)

X_train, y_train, train_paths = extract_features_from_directory(
    TRAIN_DIR,
    extractor,
    verbose=True
)

print(f"\n✓ Training features extracted: {X_train.shape}")
print(f"✓ Training labels: {y_train.shape}")
```

```python
# Cell 6: Extract Features from Validation Set
print("="*70)
print("EXTRACTING VALIDATION SET FEATURES")
print("="*70)

X_val, y_val, val_paths = extract_features_from_directory(
    VAL_DIR,
    extractor,
    verbose=True
)

print(f"\n✓ Validation features extracted: {X_val.shape}")
print(f"✓ Validation labels: {y_val.shape}")
```

```python
# Cell 7: Extract Features from Test Set
print("="*70)
print("EXTRACTING TEST SET FEATURES")
print("="*70)

X_test, y_test, test_paths = extract_features_from_directory(
    TEST_DIR,
    extractor,
    verbose=True
)

print(f"\n✓ Test features extracted: {X_test.shape}")
print(f"✓ Test labels: {y_test.shape}")
```

```python
# Cell 8: NaN/Inf diagnostics + sanitize

def nan_report(name, arr):
    nan_count = int(np.isnan(arr).sum())
    posinf_count = int(np.isposinf(arr).sum())
    neginf_count = int(np.isneginf(arr).sum())
    total = arr.size
    print(f"{name}: shape={arr.shape}  NaN={nan_count}  +Inf={posinf_count}  -Inf={neginf_count}  (total={total})")

print("Checking extracted feature matrices for NaN/Inf...")
nan_report("X_train", X_train)
nan_report("X_val", X_val)
nan_report("X_test", X_test)

# Force sanitize as a safety-net
X_train = np.nan_to_num(X_train, nan=0.0, posinf=0.0, neginf=0.0)
X_val = np.nan_to_num(X_val, nan=0.0, posinf=0.0, neginf=0.0)
X_test = np.nan_to_num(X_test, nan=0.0, posinf=0.0, neginf=0.0)

print("After sanitize:")
nan_report("X_train", X_train)
nan_report("X_val", X_val)
nan_report("X_test", X_test)
```

```python
# Cell 8: Save Features to CSV Files
print("Saving features to CSV files...")

# Get feature names
feature_names = extractor.get_feature_names()

# Create DataFrames
df_train = pd.DataFrame(X_train, columns=feature_names)
df_train['label'] = y_train
df_train['image_path'] = train_paths

df_val = pd.DataFrame(X_val, columns=feature_names)
df_val['label'] = y_val
df_val['image_path'] = val_paths

df_test = pd.DataFrame(X_test, columns=feature_names)
df_test['label'] = y_test
df_test['image_path'] = test_paths

# Save to CSV
train_csv = FEATURES_DIR / "features_train.csv"
val_csv = FEATURES_DIR / "features_val.csv"
test_csv = FEATURES_DIR / "features_test.csv"

df_train.to_csv(train_csv, index=False)
df_val.to_csv(val_csv, index=False)
df_test.to_csv(test_csv, index=False)

print(f"\n✓ Saved training features: {train_csv} ({df_train.shape})")
print(f"✓ Saved validation features: {val_csv} ({df_val.shape})")
print(f"✓ Saved test features: {test_csv} ({df_test.shape})")

# Show file sizes
print(f"\nFile sizes:")
print(f"  Train: {train_csv.stat().st_size / 1024 / 1024:.2f} MB")
print(f"  Val: {val_csv.stat().st_size / 1024 / 1024:.2f} MB")
print(f"  Test: {test_csv.stat().st_size / 1024 / 1024:.2f} MB")
```

```python
# Cell 9: Feature Statistics Summary
print("="*70)
print("FEATURE STATISTICS SUMMARY")
print("="*70)

print(f"\nTraining Set:")
print(f"  Samples: {X_train.shape[0]}")
print(f"  Features: {X_train.shape[1]}")
print(f"  Classes: {len(np.unique(y_train))}")
print(f"  Feature range: [{X_train.min():.4f}, {X_train.max():.4f}]")
print(f"  Mean: {X_train.mean():.4f}")
print(f"  Std: {X_train.std():.4f}")

print(f"\nValidation Set:")
print(f"  Samples: {X_val.shape[0]}")
print(f"  Features: {X_val.shape[1]}")
print(f"  Classes: {len(np.unique(y_val))}")
print(f"  Feature range: [{X_val.min():.4f}, {X_val.max():.4f}]")
print(f"  Mean: {X_val.mean():.4f}")
print(f"  Std: {X_val.std():.4f}")

print(f"\nTest Set:")
print(f"  Samples: {X_test.shape[0]}")
print(f"  Features: {X_test.shape[1]}")
print(f"  Classes: {len(np.unique(y_test))}")
print(f"  Feature range: [{X_test.min():.4f}, {X_test.max():.4f}]")
print(f"  Mean: {X_test.mean():.4f}")
print(f"  Std: {X_test.std():.4f}")

# Class distribution
print(f"\nClass Distribution (Training):")
unique, counts = np.unique(y_train, return_counts=True)
for cls, cnt in zip(unique, counts):
    print(f"  Class {cls:2d}: {cnt:4d} samples ({cnt/len(y_train)*100:5.2f}%)")
```

```python
# Cell 10: Visualize Feature Distributions
# Select a few representative features to visualize
feature_indices = [
    0,   # HOG feature
    extractor.feature_dims['hog'],  # First color histogram feature
    extractor.feature_dims['hog'] + extractor.feature_dims['color_hist'],  # First LBP feature
    -10  # One of the edge features
]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for idx, feat_idx in enumerate(feature_indices):
    ax = axes[idx]
    
    # Plot histogram for this feature
    ax.hist(X_train[:, feat_idx], bins=50, alpha=0.7, edgecolor='black')
    ax.set_xlabel('Feature Value', fontsize=10)
    ax.set_ylabel('Frequency', fontsize=10)
    ax.set_title(f'Feature {feat_idx}: {feature_names[feat_idx]}', fontsize=11)
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(PROJECT_ROOT / "outputs" / "plots" / "feature_distributions.png", 
            dpi=150, bbox_inches='tight')
plt.show()

print("✓ Feature distribution plot saved")
```

```python
# Cell 11: Feature Correlation Analysis (Sample)
# Analyze correlation between first 100 features (too many to visualize all)
print("Computing feature correlations (sample of 100 features)...")

# Sample features evenly across feature types
sample_indices = np.linspace(0, X_train.shape[1]-1, 100).astype(int)
X_sample = X_train[:, sample_indices]
sample_names = [feature_names[i] for i in sample_indices]

# Compute correlation matrix
corr_matrix = np.corrcoef(X_sample.T)

# Plot correlation heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, 
            xticklabels=False, 
            yticklabels=False,
            cmap='coolwarm', 
            center=0,
            vmin=-1, vmax=1,
            square=True)
plt.title('Feature Correlation Matrix (Sample of 100 features)', fontsize=14)
plt.tight_layout()
plt.savefig(PROJECT_ROOT / "outputs" / "plots" / "feature_correlation.png", 
            dpi=150, bbox_inches='tight')
plt.show()

# Report highly correlated features
high_corr = np.where((np.abs(corr_matrix) > 0.9) & (corr_matrix != 1.0))
print(f"\nFound {len(high_corr[0])//2} pairs of highly correlated features (|r| > 0.9)")
print("This is expected and will be addressed in feature selection.")
```
